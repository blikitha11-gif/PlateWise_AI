This notebook demonstrates exploratory data analysis of historical canteen preparation, consumption, and discarded food data.

```python
import pandas as pd
import numpy as np
from pathlib import Path

csv_path = Path('../data/sample_meals.csv')
df = pd.read_csv(csv_path)
df.head()
```

## 1. Summary Statistics & Data Quality Check

```python
print("Shape:", df.shape)
print("Missing values:\n", df.isnull().sum())
df.describe()
```

## 2. Demand by Day of Week and Menu Item

```python
pivot_demand = df.pivot_table(index='menu_item', columns='day_of_week', values='meals_sold', aggfunc='mean').round(1)
pivot_demand
```

## 3. Food Waste Analysis (Discarded kg & Unsold Portions)

```python
df['unsold'] = df['meals_prepared'] - df['meals_sold']
waste_by_item = df.groupby('menu_item').agg(
    total_prepared=('meals_prepared', 'sum'),
    total_sold=('meals_sold', 'sum'),
    total_unsold=('unsold', 'sum'),
    discarded_kg=('food_discarded_kg', 'sum')
).reset_index()
waste_by_item
```

## 4. Outlier Investigation
Confirming if demand spikes correspond to special campus events.

```python
q1 = df['meals_sold'].quantile(0.25)
q3 = df['meals_sold'].quantile(0.75)
iqr = q3 - q1
outliers = df[(df['meals_sold'] < q1 - 1.5 * iqr) | (df['meals_sold'] > q3 + 1.5 * iqr)]
outliers[['date', 'menu_item', 'meals_sold', 'special_event', 'is_holiday']]
```
